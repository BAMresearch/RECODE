````python
class Measurement:
    def __init__(self, values=[], units="cm", sample_id="", operator="Max Mustermann", location=""):
        self.values = values
        self.units = units
        self.sample_id = sample_id
        self.operator = operator
        self.location = location
        print("Measurement object created")

        if not self.are_values_valid():
            print("A value is negative")

        if not self.is_location_valid():
            print("Location is not valid")

    def are_values_valid(self):
        for value in self.values:
            if value >= 0:
                continue
            return False
        return True

    def is_location_valid(self):
        if self.location in ["BAM UE", "BAM AH", "BAM TTS"]:
            return True
        else:
            return False

    def calculate_average(self):
        return sum(self.values) / len(self.values)
````
