```python
# Load the necessary packages

import pandas as pd
import numpy as np
from pathlib import Path
import os
import matplotlib.pyplot as plt
```

```python
# set home directory
home_dir = Path.home()

# set path to folder that contains the unzipped .csv files of all spectra
spectra_folder = home_dir / "Miniconda3/envs/jlab/Scripts/recode/synth_spectra/" #this path potentially needs to be adapted
```

```python
# we now want to read all spectra and create a single pandas dataframe from them

# to iterate over the individual paths of all .csv-files, we need a list with all their names first
# We can either create one by hand, or use the os package to generate them automatically:

spectra_in_folder = os.listdir(spectra_folder) #listdir -> lists all files in the provided folder path
print(spectra_in_folder)
```

### There are multiple options to load and create a single dataframe.

#### We will show two. No problem if you have a different one but see identical results!

```python
# First Option

# loop to create a list with dataframes, which are put together after the loop
spectra = []

for spectrum in spectra_in_folder:
    # this if statement ensures that only spectra are loaded (in case there are other .csv files in your folder)
    if spectrum.endswith(".csv") and spectrum.startswith("spectrum"):
        selected_spectrum = pd.read_csv(spectra_folder / spectrum)
        spectra.append(selected_spectrum)
        
# concatenate the list afterwards        
df_v1 = pd.concat(spectra, ignore_index=True)
# look at the first rows to verify 
df_v1.head()
```

```python
# Second Option

# create an empty dataframe. Then loop over the .csv and concatenate each new spectral dataframe at every step 
df_v2 = pd.DataFrame()

for spectrum in spectra_in_folder:
    # this if statement ensures that only spectra are loaded (in case there are other .csv files in your folder)
    if spectrum.endswith(".csv") and spectrum.startswith("spectrum"):
        # read the spectrum
        spec_to_add = pd.read_csv(spectra_folder / spectrum)
        # concatenate it to the existing dataframe
        df_v2 = pd.concat([df_v2,spec_to_add], ignore_index=True)
        
# look at the first rows to verify        
df_v2.head()
```

#### Set all values below 0 to 0.

##### There are again multiple ways to do this. The simplest one is shown here

```python
# Create a copy to not mess with the original dataframe
df_v1_copy = df_v1.copy()

df_v1_copy[df_v1_copy<0] = 0

# df_v1_copy<0 -> this statement returns a boolean that is true for all entries below 0, and false for all entries above 0
# the boolean has the same shape as our orinigal dataframe. If we use the [] to index the entries in our dataframe, every
# entrie that is true will be set to the value after the = sign. Every entry that is false will remain the same. 

df_v1_copy.head()
```

#### I uploaded a second file that contains element concentrations of the same samples that produced the first 25 spectra (i.e. one concentration per spectra)
We now want to check, if the peaks/the intensity values between 260 and 310 are correlated with the concentrations.

Thereto, 
- Download and open the .csv with the concentrations from the RECODE repository (folder: project)
- For every spectrum, sum the intensities of all bands between 260 and 310
- Plot the intesities of the 25 spectra vs. their concentrations

```python
conc = pd.read_csv(home_dir / "Miniconda3/envs/jlab/Scripts/recode/synth_spectra/concentrations.csv")
conc.head()
```

```python
# We use the numpy function np.sum to summarize over the columns (axis=1)
peak_sum = np.sum(df_v1_peak, axis=1)
```

```python
# We plot the first 25 sums vs. their corresponding concentrations
plt.scatter(peak_sum.iloc[0:25], conc)
```

It looks like there might be a correlation. Use the sklearn package to fit a linear regression model to the data (more specifically, I recommend from sklearn.linear_model import LinearRegression)
If sklearn is not installed in your environment yet, have a look at session 6 or 7, there is a hint in the respective notebook.

Hint: the linear regression function likes numpy arrays a lot. You might need to reshape input and output though.

```python
from sklearn.linear_model import LinearRegression

# first, we need to reshape the input to numpy arrays
peak_sum_np = np.array(peak_sum.iloc[0:25])
conc_np = np.array(conc)

# we create a LinearRegression object
lin_reg = LinearRegression()
# The .fit function of this object is now used to fit th concentrations to the summarized peaks
# We need to reshape the arrays beforehand
lin_reg.fit(peak_sum_np.reshape(-1, 1), conc_np.reshape(-1, 1))
```

Now use this model on the intensity sums of the remaining 5 spectra and plot the predicted concentrations using a different color as the original 25 points.

Bonus: plot the regression line as well.

```python
pred_remaining_conc = lin_reg.predict(np.array(peak_sum.iloc[25:]).reshape(-1,1))
```

```python
# to plot the line, we extract intercept and slope:
slope = lin_reg.coef_[0]
intercept = lin_reg.intercept_

# and create the line ourself: 
x_line = np.linspace(peak_sum_np.min(), peak_sum_np.max(), 200)
y_line = slope * x_line + intercept
```

```python
plt.scatter(peak_sum.iloc[0:25], conc, color="black")
plt.scatter(peak_sum.iloc[25:], pred_remaining_conc, color="red")
plt.plot(x_line, y_line)
```

But wait, what is that? We actually found the remaining five concentrations on our hard drive! 

Load the check_predictions.csv file and compare the measurements with the predicted results.

The Root Mean Squared Error is a basic function to compare the predictions with the real values.
Luckily, we do not have to implement it ourselves and can just use the 

```python
from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(pred_remaining_conc, conc_real)
print(rmse)
```
